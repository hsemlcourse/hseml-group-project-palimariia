```python
import json
import joblib
import numpy as np
```

## загрузка данных

```python
X_train_scaled = np.load('data/X_train_scaled.npy')
y_train = np.load('data/y_train.npy')
scaler = joblib.load('data/scaler.pkl')

X_test_scaled = np.load('data/X_test_scaled.npy')
y_test = np.load('data/y_test.npy')
X_val_scaled = np.load('data/X_val_scaled.npy')
y_val = np.load('data/y_val.npy')
```

## загрузка модели

```python
xgb_model = joblib.load('models/xgb_optuna_tuned.joblib')
cat_model = joblib.load('models/catboost_optuna_tuned.joblib')
hist_model = joblib.load('models/histgb_optuna_tuned.joblib')

with open('models/ensemble_weights.json', 'r') as f:
    weights = json.load(f)

probs_xgb = xgb_model.predict_proba(X_val_scaled)[:, 1]
probs_cat = cat_model.predict_proba(X_val_scaled)[:, 1]
probs_hist = hist_model.predict_proba(X_val_scaled)[:, 1]

total_w = weights['xgb_weight'] + weights['cat_weight'] + weights['hist_weight']
final_probs = (weights['xgb_weight']*probs_xgb + weights['cat_weight']*probs_cat + weights['hist_weight']*probs_hist) / total_w
```
